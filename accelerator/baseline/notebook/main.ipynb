### Step 9: Run Hardware Emulation

This step has to be performed manually in a seperate terminal. This section will guide you through the steps:

Start the hardware emulation in a new interactive terminal

``` bash
make run
```

You will now see the following things happening:

- QEMU will startup, as well as the SystemC AIE Simulator and the RTL Simulator
- QEMU will boot the Linux on the PS
- Our test application will be automatically run 

When the application finished successfully, copy the files from the QEMU simulator. The easiest way to do this is to use `scp` over the provided port `2222`. The password is `root`
inside the virtual machine. 

```bash
mkdir -p build/hw_emu
scp -P 2222 -o StrictHostKeyChecking=no -o UserKnownHostsFile=/dev/null  root@localhost:/mnt/* build/hw_emu
```

After you have copied all your results, you may close the simulator by pressing the combination `CTRL` + `A`, followed by `X`.

### Step 10: Compare Hardware Emulation vs QKeras Reference

````python
from IPython.display import display
import h5py
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

from deeploy.verification.histplot import histplot
from deeploy.util.types import QFormat,Quantizer

workdir = Path('../build')
testpath = workdir / "hw_emu" / "output.h5"
referencepath = workdir / "../data" / "qkeras_reference.h5"

with h5py.File(testpath, "r") as f:
    test_predictions = np.array(f["int_features"])[:100,:,:]

test_energy = test_predictions[..., 0]
test_tposx  = test_predictions[..., 1]
test_tposy  = test_predictions[..., 2]
test_tposz  = test_predictions[..., 3]
test_signal = test_predictions[..., 4]

with h5py.File(referencepath, "r") as f:
    reference_predictions = np.array(f["predictions"])[:100,:,:]

reference_energy = reference_predictions[..., 1]
reference_tposx  = reference_predictions[..., 2]
reference_tposy  = reference_predictions[..., 3]
reference_tposz  = reference_predictions[..., 4]
reference_signal = reference_predictions[..., 5]

fig = histplot(test_energy, reference_energy, workdir / f"2d_histogram_energy","Test", "Reference", f"Energy Predictions", dpi=100,save_pdf=False)
plt.close(fig)
fig = histplot(test_tposx, reference_tposx, workdir / f"2d_histogram_tposx","Test", "Reference", f"Position X Predictions", dpi=100,save_pdf=False)
plt.close(fig)
fig = histplot(test_tposy, reference_tposy, workdir / f"2d_histogram_tposy","Test", "Reference", f"Position Y Predictions", dpi=100,save_pdf=False)
plt.close(fig)
fig = histplot(test_tposz, reference_tposz, workdir / f"2d_histogram_tposz","Test", "Reference", f"Position Z Predictions", dpi=100,save_pdf=False)
plt.close(fig)
fig = histplot(test_signal, reference_signal, workdir / f"2d_histogram_signal","Test", "Reference", f"Signal Predictions", dpi=100,save_pdf=False)
plt.close(fig)
````
